```python
import matplotlib.pyplot as plt

plt.style.use("dark_background")
```

```python
def funcion_distribucion(x):
    if x < 0:
        return 0
    elif x < 1:
        return 0.5
    else:
        return 1
```

```python
probabilidad_cara = funcion_distribucion(1) - funcion_distribucion(0)
probabilidad_ceca = funcion_distribucion(0) - funcion_distribucion(-1)

print(probabilidad_cara)
print(probabilidad_ceca)
```

```python
x_vals = (-1, 0, 1, 2)
y_vals = [funcion_distribucion(x) for x in x_vals]

plt.step(x_vals, y_vals, where = "post", label = "Funcion de distribucion")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("funcion distribucion de la variable aleatoria x")
plt.legend()
plt.grid(True)
plt.show()
```
